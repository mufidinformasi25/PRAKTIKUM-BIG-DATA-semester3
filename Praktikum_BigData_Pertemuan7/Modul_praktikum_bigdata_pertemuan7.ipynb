## 7.1 Studi Kasus: Laporan Kinerja Triwulan Jaringan Toko Retail

**Konteks:** Sebuah perusahaan retail memiliki **4 cabang toko** di berbagai kota. Manajemen pusat meminta laporan kinerja triwulan (3 bulan terakhir) yang mencakup performa penjualan per toko, produk terlaris di masing-masing toko, dan kontribusi tiap manajer cabang. Seluruhnya harus diproses dari data mentah yang tersebar di tiga sumber berbeda.

### Persiapan: Membuat SparkSession dan Tiga Sumber Data

```python
from pyspark.sql import SparkSession
from pyspark.sql.functions import col, sum as spark_sum, avg, count, row_number, when
from pyspark.sql.window import Window
import numpy as np
import pandas as pd
import json

spark = SparkSession.builder \
    .appName("Pertemuan7-StudiKasusTerintegrasi") \
    .master("local[*]") \
    .getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

print("SparkSession siap. Versi Spark:", spark.version)
```

```python
np.random.seed(77)

# SUMBER 1: Data toko/cabang (JSON), tabel referensi kecil
toko_list = ["Toko Magelang", "Toko Yogyakarta", "Toko Semarang", "Toko Solo"]
data_toko = [
    {"toko_id": i + 1, "nama_toko": t, "kota": t.split()[1], "manager": np.random.choice(["Rani", "Joko", "Sari", "Bayu"])}
    for i, t in enumerate(toko_list)
]
with open("data_toko.json", "w") as f:
    for t in data_toko:
        f.write(json.dumps(t) + "\n")

# SUMBER 2: Data produk (CSV), tabel referensi kecil
data_produk = [
    {"product_id": i, "nama_produk": f"Produk-{i}",
     "kategori": np.random.choice(["Elektronik", "Fashion", "Makanan"]),
     "harga": int(np.random.choice([50000, 100000, 200000]))}
    for i in range(1, 21)
]
pd.DataFrame(data_produk).to_csv("data_produk.csv", index=False)

# SUMBER 3: Data transaksi triwulan (CSV), data utama, jauh lebih besar
n = 5000
data_transaksi = pd.DataFrame({
    "trx_id": [f"T{i}" for i in range(n)],
    "toko_id": np.random.randint(1, 5, size=n),
    "product_id": np.random.randint(1, 21, size=n),
    "qty": np.random.randint(1, 6, size=n),
})
data_transaksi.to_csv("data_transaksi_triwulan.csv", index=False)

print("Ketiga sumber data berhasil dibuat:")
print(f"- data_toko.json              : {len(data_toko)} baris")
print(f"- data_produk.csv             : {len(data_produk)} baris")
print(f"- data_transaksi_triwulan.csv : {len(data_transaksi)} baris")
```

## 7.2 Tahap EXTRACT: Membaca dari HDFS

```python
df_transaksi = spark.read.csv(
    "hdfs://localhost:9000/home/mufid/praktikum-bigdata0046/pertemuan7/raw/data_transaksi_triwulan.csv",
    header=True, inferSchema=True
)
df_toko = spark.read.json("hdfs://localhost:9000/home/mufid/praktikum-bigdata0046/pertemuan7/raw/data_toko.json")
df_produk = spark.read.csv(
    "hdfs://localhost:9000/home/mufid/praktikum-bigdata0046/pertemuan7/raw/data_produk.csv",
    header=True, inferSchema=True
)

print("Transaksi:", df_transaksi.count(), "| Toko:", df_toko.count(), "| Produk:", df_produk.count())
```

## 7.3 Tahap TRANSFORM

### 7.3.1 Menggabungkan Ketiga Sumber (Join)

```python
df_gabungan = df_transaksi.join(df_toko, on="toko_id", how="inner") \
                           .join(df_produk, on="product_id", how="inner")

df_gabungan = df_gabungan.withColumn("total_penjualan", col("qty") * col("harga"))
df_gabungan.show(5)
```

### 7.3.2 Window Function: Top 3 Produk Terlaris per Toko

```python
# Meringkas total penjualan per kombinasi toko + produk terlebih dahulu
ringkasan_produk_toko = df_gabungan.groupBy("nama_toko", "nama_produk").agg(
    spark_sum("total_penjualan").alias("total_penjualan")
)

# Menerapkan window function untuk mengambil top 3 produk pada setiap toko
window_toko = Window.partitionBy("nama_toko").orderBy(col("total_penjualan").desc())
top3_produk = ringkasan_produk_toko.withColumn("peringkat", row_number().over(window_toko)) \
                                    .filter(col("peringkat") <= 3)

top3_produk.orderBy("nama_toko", "peringkat").show(20)
```

### 7.3.3 Spark SQL: Kontribusi per Manajer Cabang

```python
df_gabungan.createOrReplaceTempView("gabungan")

kontribusi_manager = spark.sql('''
    SELECT kota, manager, SUM(total_penjualan) AS total_penjualan
    FROM gabungan
    GROUP BY kota, manager
    ORDER BY total_penjualan DESC
''')
kontribusi_manager.show()
```

### Verifikasi Akhir: Ringkasan Eksekutif

Sebagai penutup, disusun satu tabel ringkasan yang layak disebut *executive summary*, persis seperti yang akan diminta manajemen dalam pekerjaan nyata.

```python
df_final = spark.read.parquet("hdfs://localhost:9000/home/mufid/praktikum-bigdata0046/pertemuan7/processed/laporan_triwulan")

ringkasan_eksekutif = df_final.groupBy("nama_toko", "kota", "manager").agg(
    spark_sum("total_penjualan").alias("total_penjualan_triwulan"),
    count("trx_id").alias("jumlah_transaksi"),
    avg("total_penjualan").alias("rata_rata_nilai_transaksi")
).orderBy(col("total_penjualan_triwulan").desc())

ringkasan_eksekutif.show()
```

Satu alur analisis Big Data yang utuh telah selesai: dari data mentah tersebar di berbagai sumber dan format, tersimpan di HDFS, diproses dengan Spark (join, window function, Spark SQL), hingga tersimpan rapi sebagai laporan akhir berformat Parquet yang siap dipakai tim lain atau divisualisasikan di dashboard.

## Latihan Mandiri

Cell pembuatan `SparkSession` dan ketiga DataFrame (`df_transaksi`, `df_toko`, `df_produk`, dibaca dari HDFS) dijalankan ulang sebelum mengerjakan latihan berikut.

```python
# Persiapan ulang untuk latihan
spark = SparkSession.builder.appName("Latihan7").master("local[*]").getOrCreate()
spark.sparkContext.setLogLevel("ERROR")

df_transaksi = spark.read.csv("hdfs://localhost:9000/home/mufid/praktikum-bigdata0046/pertemuan7/raw/data_transaksi_triwulan.csv", header=True, inferSchema=True)
df_toko = spark.read.json("hdfs://localhost:9000/home/mufid/praktikum-bigdata0046/pertemuan7/raw/data_toko.json")
df_produk = spark.read.csv("hdfs://localhost:9000/home/mufid/praktikum-bigdata0046/pertemuan7/raw/data_produk.csv", header=True, inferSchema=True)
df_gabungan = df_transaksi.join(df_toko, on="toko_id").join(df_produk, on="product_id")
df_gabungan = df_gabungan.withColumn("total_penjualan", col("qty") * col("harga"))
print("Siap untuk latihan. Total baris:", df_gabungan.count())
```

**Soal 1.** Menggunakan Spark SQL, tampilkan **kategori produk** dengan total `qty` (jumlah unit terjual, bukan nilai rupiah) tertinggi secara keseluruhan.

```python
# Jawaban Soal 1 di sini
# Penjelasan: total qty (unit terjual) dijumlahkan per kategori dengan Spark SQL,
# diurutkan dari terbesar, lalu diambil satu baris teratas dengan LIMIT 1.
df_gabungan.createOrReplaceTempView("gabungan")

spark.sql('''
    SELECT kategori, SUM(qty) AS total_qty
    FROM gabungan
    GROUP BY kategori
    ORDER BY total_qty DESC
    LIMIT 1
''').show()
```

**Soal 2.** Menggunakan window function, tentukan **toko dengan penjualan tertinggi** untuk **masing-masing kategori produk** (top 1 saja, gunakan `row_number()`).

```python
# Jawaban Soal 2 di sini
# Penjelasan: total penjualan diringkas per kombinasi kategori + toko, lalu window function
# row_number() memberi peringkat di dalam tiap kategori. Peringkat 1 adalah toko tertinggi.
ringkasan_kategori_toko = df_gabungan.groupBy("kategori", "nama_toko").agg(
    spark_sum("total_penjualan").alias("total_penjualan")
)

window_kategori = Window.partitionBy("kategori").orderBy(col("total_penjualan").desc())
toko_terbaik = ringkasan_kategori_toko.withColumn("peringkat", row_number().over(window_kategori)) \
                                      .filter(col("peringkat") == 1)

toko_terbaik.orderBy("kategori").show()
```

**Soal 3.** Simpan hasil `ringkasan_eksekutif` (dari Subbab 7.4 di atas, dibuat ulang) ke HDFS sebagai Parquet **tanpa partisi**, pada path `/home/mufid/praktikum-bigdata0046/latihan7/ringkasan_eksekutif`.

> **Catatan path:** Instruksi modul aslinya memakai path `/user/mahasiswa/latihan7/ringkasan_eksekutif`, namun disesuaikan menjadi `/home/mufid/praktikum-bigdata0046/latihan7/ringkasan_eksekutif` mengikuti konvensi direktori kerja pribadi yang sudah dipakai konsisten sejak Pertemuan 3.

```python
# Jawaban Soal 3 di sini
# Penjelasan: ringkasan_eksekutif dibuat ulang dari df_gabungan, lalu ditulis ke HDFS
# sebagai Parquet tanpa partisi (tidak memakai partitionBy).
ringkasan_eksekutif = df_gabungan.groupBy("nama_toko", "kota", "manager").agg(
    spark_sum("total_penjualan").alias("total_penjualan_triwulan"),
    count("trx_id").alias("jumlah_transaksi"),
    avg("total_penjualan").alias("rata_rata_nilai_transaksi")
).orderBy(col("total_penjualan_triwulan").desc())

!hdfs dfs -mkdir -p /home/mufid/praktikum-bigdata0046/latihan7

ringkasan_eksekutif.write.mode("overwrite").parquet(
    "hdfs://localhost:9000/home/mufid/praktikum-bigdata0046/latihan7/ringkasan_eksekutif"
)

!hdfs dfs -ls /home/mufid/praktikum-bigdata0046/latihan7/ringkasan_eksekutif

# Verifikasi: baca kembali dari HDFS
spark.read.parquet("hdfs://localhost:9000/home/mufid/praktikum-bigdata0046/latihan7/ringkasan_eksekutif").show()
```

**Soal 4 (Refleksi menyeluruh).** Sebagai persiapan UTS, jelaskan dalam 4 sampai 5 kalimat: bagaimana **HDFS, Spark, dan format Parquet saling melengkapi** dalam satu ekosistem Big Data?

**Jawaban:**

HDFS berperan sebagai *distributed storage* yang memecah dan mereplikasi data ke banyak node, sehingga data mentah tersimpan aman dan bisa dibaca secara paralel. Spark menjadi mesin *in memory processing* yang mengambil data dari HDFS, lalu menjalankan join, window function, dan Spark SQL secara terdistribusi tanpa harus terus menulis hasil antara ke disk. Parquet adalah *columnar format* yang terkompresi dan menyimpan skema di dalam berkas, sehingga hasil olahan Spark yang ditulis kembali ke HDFS lebih hemat ruang dan cepat dibaca. Dengan *partitioning*, misalnya per kategori, Spark cukup membuka folder yang relevan saja (*partition pruning*). Ketiganya membentuk satu siklus utuh: simpan mentah di HDFS, olah dengan Spark, lalu simpan hasilnya sebagai Parquet di HDFS agar siap dipakai analisis berikutnya.
